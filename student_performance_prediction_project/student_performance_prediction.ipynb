## 3. Try a different student
Change the three values below. Keep them within the ranges specified in the README.

```python
from main import generate_dataset, FEATURES
import pandas as pd

# Re-create and train the models for a custom prediction
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
import numpy as np

data = generate_dataset()
X, y = data[FEATURES], data['Final_Marks']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
linear = LinearRegression().fit(X_train, y_train)
forest = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train, y_train)
models = [linear, forest]
best_model = min(models, key=lambda model: mean_absolute_error(y_test, model.predict(X_test)))

new_student = pd.DataFrame({
    'Study_Hours': [7],
    'Attendance': [90],
    'Previous_Marks': [80]
})
estimate = float(np.clip(best_model.predict(new_student)[0], 0, 100))
print(f'Estimated final marks: {estimate:.2f}')
```

## 5. Interpretation and limitations
MAE, MSE and RMSE measure prediction error; lower values are generally better. R² compares predictions with a mean-value baseline. Results are only illustrative because the dataset is synthetic. Real-world use would require authorized real data, independent validation, privacy safeguards, fairness evaluation and human oversight.
